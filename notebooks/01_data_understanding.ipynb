### Column By Column Understanding
---
- **PatientId** : Helps in identifying each patient with the help of a unique number which is beneficial for tracking a specific case , not usable for the prediction.
- **AppointmentID** : Plays as a Unique identifier for the appointment day.
- **Gender** : It is a categorical feature which basically describes the gender of the patient.
- **ScheduledDay** : Represents the date and time when the appointment was booked by the patient.
- **AppointmentDay** : Represents the actual date on which the patient was expected to visit the hospital, known before the
appointment and safe for time-based feature engineering.
- **Age** - A numerical feature represents the exact age of the particular feature and is important one because age is related to different behaviors.
- **Neighbourhood** - Represents the area where the patient resides, categorical feature that may reflect accessibility or socio-economic factors.
- **Scholarship** - Indicates whether the patient is enrolled in a social welfare assistance program.
- **``Hipertension`` ``Diabetes`` ``Alcoholism`` ``Handcap``** - Binary medical condition indicators where 0 represents absence and 1 represents presence of the condition.
- **SMS-recieved** - Indicates whether the patient received an SMS reminder for the appointment, potentially influences patient attendance behavior.
- **No-show** - It is the target variable which is to be predicted and represents that patient went to the hospital on appointment day or not.  

## Features Role Classification
---

### Identifiers
- Patient Id
- Appointment Id

These represents the unique identity of each patient which distincts them from the others. They will not be used for prediction.

### Patient Attributes
- Age
- Gender
- Neighbourhood
- Scholarship
- Hipertension
- Diabetes
- Alcohol
- Handcap

These are the patient's Demographics and medical history which can be essential in knowung the behaviour before the appointment day.

### Time-Based Features
- AppointmentDay
- ScheduledDay

These define the date of booking and date of appointment, these should be handled carefully to prevent data leakage.

### Communication Feature
- SMS-recieved

Represents whether the patient recieved the message prior to the environment or not, can directly affect the attendance.

### Target Variable
- No-show

This is the main feature which is to be predicted and describes whether patient went to the hospital on the appointment date or not, it will not be considered as the input feature.
